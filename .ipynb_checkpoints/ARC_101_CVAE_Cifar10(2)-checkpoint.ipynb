Here’s a clean, self-contained PyTorch implementation of a **Conditional Variational Autoencoder (CVAE)** trained with the **ELBO** (reconstruction + KL) objective. It uses MNIST digits as the condition (one-hot class labels), but you can swap in your own condition vectors easily.

### How it works (quick notes)

* **Conditioning**: we one-hot encode the label `c` and concatenate it to the image vector for both encoder (`[x, c]`) and decoder (`[z, c]`).
* **Reparameterization**: `z = μ + σ ⊙ ε`, with `ε ~ N(0, I)`.
* **ELBO**: `loss = BCEWithLogits(x_logits, x) + β * KL(q(z|x,c) || N(0,I))`, where the BCE sums over pixels and we average over the batch. `β=1` is the standard CVAE; adjust for β-VAE or do KL annealing if you like.

### Tweaks you might want

* Replace Bernoulli with Gaussian likelihood for real-valued images: use MSE (or Gaussian NLL) on `decoder` outputs without logits (i.e., apply `sigmoid` inside the model and predict means).
* Change architectures to ConvNets for higher-res images (swap the MLPs for Conv/Deconv blocks).
* KL annealing: ramp `beta` from 0 → 1 over the first few epochs.



Awesome—let’s switch your CVAE to **CIFAR-10** with solid preprocessing, normalization, and data augmentation. Below is a clean, end-to-end PyTorch script that:

* downloads CIFAR-10
* applies standard augmentations (crop/flip/jitter/erasing) to the train set
* normalizes with CIFAR-10 channel stats
* uses a **convolutional CVAE** (32×32×3) with an **ELBO** (MSE recon in normalized space + KL)
* conditions on class labels via one-hot vectors (concat to the latent and to the encoder input)
* saves a denormalized sample grid for quick inspection

> Tip: because we train on **normalized** images, we compute the reconstruction in normalized space with **MSE**. For visuals, we **denormalize** and clamp to \[0,1].


````python
# cvae_cifar10.py
# ------------------------------------------------------------
# Convolutional CVAE on CIFAR-10 with proper preprocessing and
# data augmentation. ELBO = MSE recon (normalized space) + KL.
# ------------------------------------------------------------
import random
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
import torchvision
from torchvision import datasets, transforms, utils as vutils
from IPython.display import display
import matplotlib.pyplot as plt
import numpy as np
from sklearn.decomposition import PCA
# import torchvision.transforms as T
# from IPython.display import display
````

````python
# -----------------------
# Config / Utils
# -----------------------
CIFAR10_MEAN = (0.4914, 0.4822, 0.4465)
CIFAR10_STD  = (0.2023, 0.1994, 0.2010)

def set_seed(seed: int = 42):
    random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

def one_hot(y: torch.Tensor, num_classes: int) -> torch.Tensor:
    return F.one_hot(y, num_classes=num_classes).float()

def get_transforms(train: bool = True):
    if train:
        # light but common CIFAR-10 recipe
        return transforms.Compose([
                # transforms.RandomCrop(32, padding=4, padding_mode='reflect'),
                transforms.RandomHorizontalFlip(p=0.5),
                transforms.ToTensor(),
                transforms.Normalize(CIFAR10_MEAN, CIFAR10_STD),
            ])
    else:
        return transforms.Compose([
            transforms.ToTensor(),
            transforms.Normalize(CIFAR10_MEAN, CIFAR10_STD),
        ])

@torch.no_grad()
def denormalize(x: torch.Tensor):
    # x: (B,3,H,W) in normalized space -> [0,1] for saving
    mean = torch.tensor(CIFAR10_MEAN, dtype=x.dtype, device=x.device).view(1,3,1,1)
    std  = torch.tensor(CIFAR10_STD,  dtype=x.dtype, device=x.device).view(1,3,1,1)
    x = x * std + mean
    return x.clamp(0, 1)
````

````python
from torchvision import models

IMAGENET_MEAN = (0.485, 0.456, 0.406)
IMAGENET_STD  = (0.229, 0.224, 0.225)

class VGGPerceptual(nn.Module):
    """
    Outputs activations from selected VGG19 layers.
    Uses pretrained ImageNet weights; frozen in eval mode.
    """
    def __init__(self, layers=(3, 8, 17, 26), upsample_to_224=True):
        super().__init__()
        vgg = models.vgg19(weights=models.VGG19_Weights.IMAGENET1K_V1).features
        self.slices = nn.ModuleList()
        last_idx = 0
        for idx in layers:
            self.slices.append(nn.Sequential(*[vgg[i] for i in range(last_idx, idx + 1)]))
            last_idx = idx + 1
        for p in self.parameters():
            p.requires_grad = False
        self.eval()
        self.layers = layers
        self.upsample_to_224 = upsample_to_224

        # register ImageNet norm buffers for speed
        mean = torch.tensor(IMAGENET_MEAN).view(1,3,1,1)
        std  = torch.tensor(IMAGENET_STD ).view(1,3,1,1)
        self.register_buffer("in_mean", mean, persistent=False)
        self.register_buffer("in_std",  std,  persistent=False)

    def preprocess(self, x_01: torch.Tensor) -> torch.Tensor:
        """
        x_01: [0,1] range (B,3,H,W) → normalize for ImageNet VGG.
        Optionally upsample to 224 for more meaningful features.
        """
        if self.upsample_to_224 and (x_01.shape[-1] != 224 or x_01.shape[-2] != 224):
            x_01 = F.interpolate(x_01, size=(224, 224), mode="bilinear", align_corners=False)
        return (x_01 - self.in_mean) / self.in_std

    @torch.no_grad()
    def _features_no_grad(self, x):
        feats = []
        h = x
        for block in self.slices:
            h = block(h)
            feats.append(h)
        return feats

    def forward(self, x_01: torch.Tensor, y_01: torch.Tensor, p=1):
        """
        Compute Lp feature reconstruction loss between x and y in [0,1].
        Returns weighted sum over selected layers (equal weights).
        """
        x = self.preprocess(x_01)
        y = self.preprocess(y_01)
        # allow gradients to flow into x (the model output) only
        feats_x = []
        h = x
        for block in self.slices:
            h = block(h)
            feats_x.append(h)
        with torch.no_grad():
            feats_y = self._features_no_grad(y)

        losses = []
        for fx, fy in zip(feats_x, feats_y):
            if p == 1:
                losses.append(F.l1_loss(fx, fy))
            else:
                losses.append(F.mse_loss(fx, fy))
        return sum(losses) / len(losses)
````

````python
# -----------------------
# Model
# -----------------------
class ConvEncoder(nn.Module):
    """
    Encoder takes image x (3x32x32) and optional tiled condition channels.
    We concatenate a broadcasted one-hot condition (10 channels) along the channel dim.
    """
    def __init__(self, in_ch: int = 3, c_dim: int = 10, z_dim: int = 128):
        super().__init__()
        self.c_dim = c_dim
        ch = in_ch + c_dim  # concat condition (tiled) to input
        self.net = nn.Sequential(
            nn.Conv2d(ch, 64, 4, 2, 1),   # 32 -> 16
            nn.BatchNorm2d(64), nn.ReLU(True),
            nn.Conv2d(64, 128, 4, 2, 1),  # 16 -> 8
            nn.BatchNorm2d(128), nn.ReLU(True),
            nn.Conv2d(128, 256, 4, 2, 1), # 8 -> 4
            nn.BatchNorm2d(256), nn.ReLU(True),
            nn.Conv2d(256, 256, 3, 1, 1),
            nn.BatchNorm2d(256), nn.ReLU(True),
        )
        self.mu = nn.Linear(256 * 4 * 4, z_dim)
        self.logvar = nn.Linear(256 * 4 * 4, z_dim)

    def forward(self, x: torch.Tensor, c: torch.Tensor):
        # c: (B,10), tile to (B,10,32,32)
        B, _, H, W = x.shape
        c_map = c.view(B, self.c_dim, 1, 1).expand(B, self.c_dim, H, W)
        h = self.net(torch.cat([x, c_map], dim=1))
        h = h.view(B, -1)
        return self.mu(h), self.logvar(h)

class ConvDecoder(nn.Module):
    """
    Decoder maps (z,c) -> x_hat in normalized space.
    We concatenate c to z (vector), then deconv up to 32x32.
    """
    def __init__(self, z_dim: int = 128, c_dim: int = 10, out_ch: int = 3):
        super().__init__()
        self.fc = nn.Linear(z_dim + c_dim, 256 * 4 * 4)
        self.net = nn.Sequential(
            nn.ConvTranspose2d(256, 256, 4, 2, 1),  # 4 -> 8
            nn.BatchNorm2d(256), nn.ReLU(True),
            nn.ConvTranspose2d(256, 128, 4, 2, 1),  # 8 -> 16
            nn.BatchNorm2d(128), nn.ReLU(True),
            nn.ConvTranspose2d(128, 64, 4, 2, 1),   # 16 -> 32
            nn.BatchNorm2d(64), nn.ReLU(True),
            nn.Conv2d(64, out_ch, 3, 1, 1)          # logits in normalized space
        )

    def forward(self, z: torch.Tensor, c: torch.Tensor):
        h = torch.cat([z, c], dim=1)
        h = self.fc(h).view(h.size(0), 256, 4, 4)
        x_hat = self.net(h)  # No activation; we use MSE to normalized targets
        return x_hat

class CVAE(nn.Module):
    def __init__(self, z_dim: int = 128, c_dim: int = 10):
        super().__init__()
        self.z_dim = z_dim
        self.c_dim = c_dim
        self.encoder = ConvEncoder(in_ch=3, c_dim=c_dim, z_dim=z_dim)
        self.decoder = ConvDecoder(z_dim=z_dim, c_dim=c_dim, out_ch=3)

    @staticmethod
    def reparameterize(mu: torch.Tensor, logvar: torch.Tensor):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def forward(self, x: torch.Tensor, c: torch.Tensor):
        mu, logvar = self.encoder(x, c)
        z = self.reparameterize(mu, logvar)
        x_hat = self.decoder(z, c)
        return x_hat, mu, logvar
````

````python
# -----------------------
# ELBO (MSE recon + KL)
# -----------------------
# def elbo_loss_mse(x_hat: torch.Tensor, x: torch.Tensor, mu: torch.Tensor, logvar: torch.Tensor, beta: float = 1.0):
#     """
#     Reconstruction in normalized space with MSE:
#       E_q[ - (1/(2σ^2)) ||x - x_hat||^2 ] ~ MSE (σ assumed 1)
#     KL(q||p) for diagonal Gaussians:
#       0.5 * sum(exp(logvar) + mu^2 - 1 - logvar)
#     We report batch means.
#     """
#     recon = F.mse_loss(x_hat, x, reduction='none')  # (B,3,32,32)
#     recon = recon.view(recon.size(0), -1).mean(dim=1).mean()  # mean per-sample, then mean batch
#     kl = -0.5 * (1 + logvar - mu.pow(2) - logvar.exp())
#     kl = kl.sum(dim=1).mean()
#     loss = recon + beta * kl
#     return loss, {"recon": recon.detach(), "kl": kl.detach()}
# def elbo_loss_mse(x_hat, x, mu, logvar, beta: float = 1.0):
#     B = x.size(0)
#     # per-sample sum over pixels/channels
#     recon = F.mse_loss(x_hat, x, reduction='none')   # (B,3,32,32)
#     recon = recon.view(B, -1).sum(dim=1).mean()      # avg over batch

#     # KL per-sample sum over z, then avg over batch
#     kl = -0.5 * (1 + logvar - mu.pow(2) - logvar.exp()).sum(dim=1).mean()

#     loss = recon + beta * kl
#     return loss, {"recon": recon.detach(), "kl": kl.detach()}
def elbo_loss_perceptual(
    x_hat, x, mu, logvar,perc,
    beta: float = 1.0,
    w_perc: float = 1.0,
    w_pix: float = 0.05,  # small stabilizer
    p: int = 1
):
    # Map to [0,1] for perceptual features
    x_01_hat = denormalize(x_hat).clamp(0,1)
    x_01     = denormalize(x).clamp(0,1)

    # Perceptual loss
    perc_loss = perc(x_01_hat, x_01, p=p)

    # Optional pixel loss stabilizer (on normalized space)
    if w_pix > 0:
        pix = F.l1_loss(x_hat, x) if p == 1 else F.mse_loss(x_hat, x)
    else:
        pix = torch.tensor(0.0, device=x.device)

    # KL divergence
    kl = -0.5 * (1 + logvar - mu.pow(2) - logvar.exp()).sum(dim=1).mean()

    recon = w_perc * perc_loss + w_pix * pix
    loss  = recon + beta * kl
    return loss, {
        "recon": recon.detach(),          # <— added so your loop keeps working
        "recon_perc": perc_loss.detach(),
        "recon_pix":  pix.detach(),
        "kl": kl.detach()
    }
````

````python
# -----------------------
# Training / Sampling
# -----------------------
def get_dataloaders(batch_size=128, data_root="./data"):
    train_tf = get_transforms(train=True)
    test_tf  = get_transforms(train=False)

    train_ds = datasets.CIFAR10(root=data_root, train=True, download=True, transform=train_tf)
    test_ds  = datasets.CIFAR10(root=data_root, train=False, download=True, transform=test_tf)

    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=0, pin_memory=True)
    test_loader  = DataLoader(test_ds,  batch_size=batch_size, shuffle=False, num_workers=0, pin_memory=True)
    return train_loader, test_loader
````

````python
# CIFAR-10 class names
classes = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

def show_one_per_class(train_loader):
    # grab a single batch
    images, labels = next(iter(train_loader))

    chosen = {}
    for img, label in zip(images, labels):
        lbl = label.item()
        if lbl not in chosen:  # first time we see this class
            chosen[lbl] = img
        if len(chosen) == len(classes):
            break

    # plot
    fig, axes = plt.subplots(1, 10, figsize=(15, 2))
    for i, ax in enumerate(axes):
        img = chosen[i]
        img = img.permute(1, 2, 0).cpu().numpy()  # C,H,W → H,W,C
        ax.imshow((img - img.min()) / (img.max() - img.min()))  # normalize for display
        ax.set_title(classes[i], fontsize=8)
        ax.axis("off")

    plt.tight_layout()
    plt.show()
````

````python
train_loader, test_loader = get_dataloaders(batch_size=128, data_root="./data")
````

````python
show_one_per_class(train_loader)
````

````python
def train(
    # batch_size: int = 128,
    train_loader,
    test_loader,
    epochs: int = 50,
    lr: float = 2e-4,
    z_dim: int = 128,
    beta: float = 1.0,
    # data_root: str = "./data",
    device: str = None,
):
    set_seed(123)
    device = device or ("cuda" if torch.cuda.is_available() else "cpu")
    # uses your existing denormalize(x) -> [0,1]
    perc = VGGPerceptual(layers=(3, 8, 17, 26), upsample_to_224=True).to(device)
    # train_loader, test_loader = get_dataloaders(batch_size=batch_size, data_root=data_root)
    model = CVAE(z_dim=z_dim, c_dim=10).to(device)
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-4)
    # lists to track reconstruction + kl divergence
    train_rec_losses = []
    train_kl_losses = []
    test_rec_losses = []
    test_kl_losses = []
    def loop(loader, train_mode: bool):
        model.train(train_mode)
        tot_loss = tot_rec = tot_kl = n = 0
        for x, y in loader:
            x = x.to(device)           # normalized tensors
            c = one_hot(y.to(device), 10)

            if train_mode:
                opt.zero_grad()

            # x_hat, mu, logvar = model(x, c)
            # loss, parts = elbo_loss_mse(x_hat, x, mu, logvar, beta=beta)
            x_hat, mu, logvar = model(x, c)
            loss, parts = elbo_loss_perceptual(
                x_hat, x, mu, logvar,perc,
                beta=beta, w_perc=1.0, w_pix=0.05, p=1
            )

            if train_mode:
                loss.backward()
                torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
                opt.step()

            b = x.size(0)
            tot_loss += loss.item() * b
            tot_rec  += parts["recon"].item() * b
            tot_kl   += parts["kl"].item() * b
            n += b
        return tot_loss / n, tot_rec / n, tot_kl / n

    for ep in range(1, epochs + 1):
        tr_loss, tr_rec, tr_kl = loop(train_loader, True)
        te_loss, te_rec, te_kl = loop(test_loader, False)
        # store only train values (or both if you want)
        train_rec_losses.append(tr_rec)
        train_kl_losses.append(tr_kl)
        # store only train values (or both if you want)
        test_rec_losses.append(te_rec)
        test_kl_losses.append(te_kl)
        print(f"Epoch {ep:03d} | "
              f"train {tr_loss:.4f} (rec {tr_rec:.4f}, kl {tr_kl:.4f}) | "
              f"val {te_loss:.4f} (rec {te_rec:.4f}, kl {te_kl:.4f})")

    return model,train_rec_losses,train_kl_losses,test_rec_losses,test_kl_losses
````

````python
@torch.no_grad()
def show_recon(model, batch, use_logits=False):
    model.eval()
    x, y = batch  # x is already normalized
    x = x.to(next(model.parameters()).device)
    mu, logvar = model.encoder(x, one_hot(y, num_classes=10).to(x.device))
    z = model.reparameterize(mu, logvar)
    x_hat = model.decoder(z, one_hot(y, num_classes=10).to(x.device))

    # Map to [0,1] correctly:
    if use_logits:           # BCEWithLogitsLoss case
        x_vis = torch.sigmoid(x_hat).cpu()
    else:                    # MSE/L1 on normalized images
        x_vis = denormalize(x_hat).cpu()

    import torchvision.utils as vutils
    import matplotlib.pyplot as plt
    grid = vutils.make_grid(torch.cat([denormalize(x).cpu(), x_vis], dim=0), nrow=x.size(0))
    plt.figure(figsize=(x.size(0), 2))
    plt.imshow(grid.permute(1, 2, 0), interpolation="nearest")
    plt.axis("off"); plt.show()
````

````python
@torch.no_grad()
def sample_grid(model: CVAE, labels, n_per_class: int = 8, out_path: str = "cifar10_cvae_samples.png", device: str = None):
    device = device or ("cuda" if torch.cuda.is_available() else "cpu")
    model.eval().to(device)

    labels = torch.as_tensor(labels, dtype=torch.long, device=device)
    c = one_hot(labels, num_classes=10).repeat_interleave(n_per_class, dim=0)

    z = torch.randn(len(labels) * n_per_class, model.z_dim, device=device)
    x_hat = model.decoder(z, c)  # normalized space
    x_vis = denormalize(x_hat).cpu()  # [0,1]
    print(x_hat.min().item(), x_hat.max().item())
    print(x_vis.min().item(), x_vis.max().item())
    grid = vutils.make_grid(x_vis, nrow=n_per_class, padding=2)
    vutils.save_image(grid, out_path)
    # Plot inline
    plt.figure(figsize=(n_per_class, len(labels)))  # adjust size
    plt.imshow(grid.permute(1, 2, 0))  # [C,H,W] → [H,W,C]
    plt.axis("off")
    plt.show()
    print(f"Saved samples to {out_path}")
````

Great question — let’s break down what this function does step by step.

This is a **sampling/visualization helper** for your trained **Conditional Variational Autoencoder (CVAE)**.

---

### 1. Decorator: `@torch.no_grad()`

* Ensures no gradients are tracked → reduces memory use and speeds up inference.
* Important since we’re not training, only sampling.

---

### 2. Device setup

```python
device = device or ("cuda" if torch.cuda.is_available() else "cpu")
model.eval().to(device)
```

* Chooses GPU if available, otherwise CPU.
* Puts the model in evaluation mode (`eval()` disables dropout, batchnorm updates, etc.).

---

### 3. Prepare conditioning labels

```python
labels = torch.as_tensor(labels, dtype=torch.long, device=device)
c = one_hot(labels, num_classes=10).repeat_interleave(n_per_class, dim=0)
```

* Converts given labels (like `[0, 1, 2, ...]`) into **one-hot vectors**.
* Each label is repeated `n_per_class` times → ensures multiple samples per class.

Example:
If `labels = [0, 1]` and `n_per_class=3`,
then `c` will be one-hot for `[0,0,0,1,1,1]`.

---

### 4. Sample latent vectors

```python
z = torch.randn(len(labels) * n_per_class, model.z_dim, device=device)
```

* Draws **random latent vectors** from a standard normal distribution.
* Shape: `(#labels × n_per_class, latent_dim)`.

---

### 5. Decode into images

```python
x_hat = model.decoder(z, c)  # normalized space
x_vis = denormalize(x_hat).cpu()  # [0,1]
```

* The decoder takes `(z, class_condition)` and generates synthetic images.
* `x_hat` is still normalized (likely in `[-1,1]`).
* `denormalize` rescales back to pixel space `[0,1]`.

---

### 6. Make a grid and save

```python
grid = vutils.make_grid(x_vis, nrow=n_per_class, padding=2)
vutils.save_image(grid, out_path)
```

* `make_grid`: arranges generated images into a nice **grid layout**.
* `nrow=n_per_class` → each row corresponds to a label/class, each column is a different random sample.
* Saves the grid to `cifar10_cvae_samples.png`.

---

### 7. Log message

```python
print(f"Saved samples to {out_path}")
```

---

✅ **In plain words:**
This function **samples new images from your CVAE**, conditioned on given labels.
For each label, it generates `n_per_class` random variations, arranges them into a grid, and saves the result as an image file.

So if you pass `labels = range(10)` with `n_per_class=8`, you’ll get a **10×8 grid** of CIFAR-10-like images, each row showing different variations of one class.

````python
@torch.no_grad()
def reconstruct_grid(model: CVAE, loader: DataLoader, device: str, k: int = 8, out_path: str = "recon_grid.png"):
    model.eval()
    x, y = next(iter(loader))
    x = x.to(device)[:k]
    y = y.to(device)[:k]
    c = one_hot(y, 10)
    x_hat, _, _ = model(x, c)
    # show original (top) and reconstruction (bottom)
    x_vis = denormalize(x).cpu()
    x_hat_vis = denormalize(x_hat).cpu()
    grid = vutils.make_grid(torch.cat([x_vis, x_hat_vis], dim=0), nrow=k, padding=2)
    vutils.save_image(grid, out_path)
    # Plot inline
    plt.figure(figsize=(10, 10))  # adjust size
    plt.imshow(grid.permute(1, 2, 0))  # [C,H,W] → [H,W,C]
    plt.axis("off")
    plt.show()
    print(f"[viz] Saved recon grid -> {out_path}")
````

````python
@torch.no_grad()
def latent_pca_plot(model: CVAE, loader: DataLoader, device: str, out_path: str = "latent_pca.png", max_points: int = 5000):
    model.eval()
    zs, ys = [], []
    seen = 0
    for x, y in loader:
        x = x.to(device)
        c = one_hot(y.to(device), 10)
        mu, logvar = model.encoder(x, c)
        z = mu  # use mean as embedding
        zs.append(z.detach().cpu())
        ys.append(y.detach().cpu())
        seen += x.size(0)
        if seen >= max_points:
            break
    Z = torch.cat(zs, dim=0).numpy()
    Y = torch.cat(ys, dim=0).numpy()

    pca = PCA(n_components=2)
    Z2 = pca.fit_transform(Z)

    plt.figure(figsize=(6,5))
    scatter = plt.scatter(Z2[:,0], Z2[:,1], c=Y, s=8, cmap="tab10")
    plt.legend(*scatter.legend_elements(), title="class", bbox_to_anchor=(1.05, 1), loc="upper left")
    plt.title("CVAE latent space (PCA to 2D)")
    plt.tight_layout()
    plt.savefig(out_path, dpi=200)
    plt.show()
    print(f"[viz] Saved latent PCA plot -> {out_path}")
````

````python
@torch.no_grad()
def latent_traversal(model: CVAE, class_id: int, dim: int, steps: int = 11, span: float = 3.0,
                     out_path: str = "latent_traversal.png", device: str = None):
    """
    Fix z ~ N(0,I), vary one dimension over linspace(-span, span) to see effect on outputs.
    """
    device = device or ("cuda" if torch.cuda.is_available() else "cpu")
    model.eval().to(device)

    c = one_hot(torch.tensor([class_id], device=device), 10).repeat(steps, 1)
    z = torch.randn(1, model.z_dim, device=device).repeat(steps, 1)
    sweep = torch.linspace(-span, span, steps, device=device)
    z[:, dim] = sweep
    x_hat = model.decoder(z, c)
    x_vis = denormalize(x_hat).cpu()
    grid = vutils.make_grid(x_vis, nrow=steps, padding=2)
    vutils.save_image(grid, out_path)
    # Plot inline
    plt.figure(figsize=(10, 10))  # adjust size
    plt.imshow(grid.permute(1, 2, 0))  # [C,H,W] → [H,W,C]
    plt.axis("off")
    plt.show()
    print(f"[viz] Saved latent traversal -> {out_path}")
````

### Specific reasons in your case

1. **MSE loss**: encourages pixel-wise averaging → blurry images.
2. **Decoder depth**: only 3 deconvolutional layers → limited capacity.
3. **KL term**: even with β < 1, it pushes latents toward Gaussian, which sacrifices fine detail.
4. **CIFAR-10**: small, diverse dataset → harder to learn crisp conditional generation with plain CVAE.

### How to improve sample quality

If your goal is *sharp conditional samples*, you’ll need one of these upgrades:

* **Architectural**: deeper decoder, ResNet-style blocks, attention.

* **Flow prior**: VAE + normalizing flows (richer latent distribution).

### What we changed (and why)

* **Transforms (train):** `RandomCrop(32,pad=4)`, `RandomHorizontalFlip`, **mild** `ColorJitter`, and `RandomErasing(p=0.25)`. These are standard for CIFAR-10 and help generalization without destabilizing the VAE.
* **Normalization:** per-channel using `CIFAR10_MEAN` / `CIFAR10_STD`. We train and reconstruct **in normalized space** and **denormalize** only for visualization.
* **Conv CVAE:** downsample 32→16→8→4 in the encoder, then mirror with transposed convs in the decoder. This is far better than MLPs for images.
* **ELBO:** `MSE + β·KL`. For natural images, MSE in normalized space is a simple, stable choice. You can anneal `β` from 0→1 across the first \~10 epochs for crisper reconstructions.

### Optional tweaks

* Replace `ColorJitter`/`RandomErasing` with `AutoAugment(policy=CIFAR10)` from `torchvision.transforms` if you want a one-liner augmentation policy.
* Try a **Bernoulli** or **logistic** likelihood if you rescale images to \[0,1] and prefer BCE-style recon; stick to MSE when training in normalized space.
* Use **Label Embeddings** (e.g., `nn.Embedding(10, 10)` or higher-dim embeddings) instead of raw one-hots; just drop them where one-hots are used now.
